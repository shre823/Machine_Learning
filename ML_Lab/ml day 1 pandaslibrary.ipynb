```python
#creating a dataframe using dictionary
# it convert unstructure data into structure formatt
data_dict = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eva'],
    'Age': [25, 30, 35, 40, 45],
    'Salary': [50000, 60000, 75000, 80000, 90000]
}

df = pd.DataFrame(data_dict)

print("DataFrame Example:")
print(df)
```

```python
# Display top and bottom rows
print(df.head(3))  # First 3 rows
print(df.tail(2))  # Last 2 rows
```

```python
# Basic info
print(df.info())   # Provides a concise summary of the DataFrame (index dtype, column dtypes, non-null values, and memory usage)
```

```python
# Summary statistics
print(df.describe())  # Generates descriptive statistics for numeric columns (mean, std, min, quartiles, max)
```

```python
# Column names
print(df.columns)    # Returns the Index of column labels
```

```python
# Shape of dataset (rows, columns)
print(df.shape)      # Returns a tuple representing the dimensionality of the DataFrame
```

```python
# Selecting one column
print(df['Name'])  # This returns a Series containing all values from the 'Name' column.
```

```python
# Selecting multiple columns
print(df[['Name', 'Salary']])  # Double brackets create a DataFrame with the specified columns.
```

```python
# Selecting by index position (using iloc)
print(df.iloc[0])    # Returns the first row as a Series (integer-location based indexing).
```

```python
print(df.iloc[1:3])  # Returns rows 1 and 2 (slice is exclusive of the end index).
```

```python
# Selecting by label (using loc)
print(df.loc[0:2, ['Name', 'Salary']])  # Selects rows 0 to 2 and the specified columns by label.
```

```python
#iloc vs loc:iloc: Integer-based indexing (position-based)Uses integer positions (0, 1, 2, ...)
#Exclusive of the end index in slices (like Python lists)
#Example: df.iloc[0:2] gets rows 0 and 1
#loc: Label-based indexing (name-based)Uses row/column 
#labelsInclusive of the end index in slices (unlike Python lists
#)Example: df.loc[0:2] gets rows with labels 0, 1, and 2
```

```python
# Add a new column
df['Tax'] = df['Salary'] * 0.10
print(df)
```

```python
# Update existing column
df['Salary'] = df['Salary'] + 5000
print(df)
```

```python
# Delete a column
df.drop('Tax', axis=1, inplace=True)
print(df)
```

```python
# Arithmetic operations
df['Bonus'] = df['Salary'] * 0.05
print(df)
```

```python
# Introduce missing values
df.loc[2, "Age"] = None
print("With Missing Data:\n", df)
```

```python
# Check for null values
print(df.isnull())
```

```python
# Fill missing values
df["Age"] = df["Age"].fillna(df["Age"].mean())
df
```

```python
# Drop rows with missing data
# df.dropna(inplace=True)
# print("After Handling Missing Data:\n", df)
```

```python
# Filter data where Age > 30
print(df[df['Age'] > 30])
```

```python
# Multiple conditions
print(df[(df['Age'] > 30) & (df['Salary'] > 60000)])
```

```python
# Sort by Salary
print(df.sort_values(by='Salary', ascending=False))
```

```python
# Sort by multiple columns
print(df.sort_values(by=['Age', 'Salary'], ascending=[True, False]))
```

```python
# Rank the salaries
df['Salary_Rank'] = df['Salary'].rank(ascending=False)
print(df)
```

```python
#grouping and aggregation



# Group by variable
# Group by example
group_data = df.groupby('Age')['Salary'].mean()
print(group_data)

# Multiple aggregations
agg_data = df.groupby('Age').agg({'Salary': ['mean', 'max', 'min']})
print(agg_data)
```

```python
#merging and joining dataframe


# Merge on 'Name' column
dept_data = pd.DataFrame({
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eva'],
    'Department': ['HR', 'IT', 'Finance', 'IT', 'HR']
})

# Merge on multiple columns
merged_df = pd.merge(df, dept_data, on='Name')
print(merged_df)
```
