```python
# Verify installation:
import tensorflow as tf
print(f"TensorFlow version: {tf.__version__}")
print(f"GPU available: {tf.config.list_physical_devices('GPU')}")
```

```python
matrix = tf.constant([[1, 2], [3, 4]])  # 2-D tensor (matrix)
print(matrix)
```

```python
scalar = tf.constant(42)  # 0-D tensor (scalar)
vector = tf.constant([1, 2, 3, 4])  # 1-D tensor (vector)
matrix = tf.constant([[1, 2], [3, 4]])  # 2-D tensor (matrix)
tensor_3d = tf.constant([[[1, 2], [3, 4]], [[5, 6], [7, 8]]])  # 3-D tensor

print("\nTensor Examples:")
print(f"Scalar shape: {scalar.shape}, value: {scalar.numpy()}")
print(f"Vector shape: {vector.shape}, values: {vector.numpy()}")
print(f"Matrix shape: {matrix.shape}")
print(f"3D Tensor shape: {tensor_3d.shape}")
```

```python
# Tensor properties

print(f"\nTensor dtype: {matrix.dtype}")
print(f"Tensor rank: {tf.rank(matrix).numpy()}")   # Number of dimensions
print(f"Tensor size: {tf.size(matrix).numpy()}")   # Total elements
```

```python
# Basic operations
a = tf.constant([[1, 2], [3, 4]])
b = tf.constant([[5, 6], [7, 8]])

addition = tf.add(a, b)  # or a + b
multiplication = tf.multiply(a, b)  # Element-wise
matmul = tf.matmul(a, b)  # Matrix multiplication

print("\nTensor Operations:")
print(f"Addition:\n{addition.numpy()}")
print(f"Element-wise multiplication:\n{multiplication.numpy()}")
print(f"Matrix multiplication:\n{matmul.numpy()}")
```

```python
# Common mathematical functions
x = tf.constant([1.0, 2.0, 3.0])
print(f"\nSquare: {tf.square(x).numpy()}")
print(f"Square root: {tf.sqrt(x).numpy()}")
print(f"Exponential: {tf.exp(x).numpy()}")
```
