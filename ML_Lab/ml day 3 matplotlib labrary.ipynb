```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Simple Line Plot

x = np.linspace(0, 10) 
y = np.sin(x)  

plt.plot(x, y)

plt.title("Simple Line Plot of sin(x)")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")

plt.show()
```

```python
# Customize Line Style, Color, and Marker

x = np.linspace(0, 10, 50)
y1 = np.sin(x)
y2 = np.cos(x)

# Plot multiple lines with customization
plt.plot(x, y1, color='red', linestyle='--', marker='o', label='sin(x)')
plt.plot(x, y2, color='blue', linestyle='-', marker='x', label='cos(x)')

plt.title("Customized Line Plot")
plt.xlabel("X values")
plt.ylabel("Y values")

# Add grid and legend
plt.grid(True)
plt.legend()

plt.show()
```

```python
# Bar Chart

categories = ['A', 'B', 'C', 'D']
values = [23, 45, 56, 78]

plt.bar(categories, values, color='orange')

plt.title("Bar Chart Example")
plt.xlabel("Categories")
plt.ylabel("Values")

plt.show()
```

```python
# Scatter Plot

# Create random data
x = np.random.rand(50)
y = np.random.rand(50)
colors = np.random.rand(50)
sizes = 1000 * np.random.rand(50)

# Create scatter plot
plt.scatter(x, y, c=colors, s=sizes, alpha=0.5, cmap='viridis') #c means color.s means size of dots.It controls bubble size.

plt.title("Scatter Plot Example")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")

plt.colorbar()  # Add color scale
plt.show()
```

```python

# Histogram

# Random data for histogram
data = np.random.randn(1000)  # 1000 random numbers

# Plot histogram
plt.hist(data, bins=30, color='green', edgecolor='black', alpha=0.7) #bins = A histogram divides data into groups (intervals).
#It must be between 0 and 1. fully visible dark solid color
#alpha = 0.7 slightly transparent most commonly used


plt.title("Histogram Example")
plt.xlabel("Data Range")
plt.ylabel("Frequency")

plt.show()
```

```python
# np.random.rand()
# Range = 0 to 1
# No negative values
# Uniform distributio

# np.random.randn()
# Mean = 0
# Can contain negative and positive values
```

```python
#Pie chart

# Data for pie chart
labels = ['Apple', 'Banana', 'Cherry', 'Mango']
sizes = [30, 20, 25, 25]
colors = ['red', 'yellow', 'pink', 'orange']

# Create pie chart
plt.pie(sizes, labels=labels, colors=colors, autopct='%1.1f%%', startangle=90, shadow=True) #autopct= It shows percentage text on slices.
                                                                                            #startangle=90 This rotates the chart.
plt.title("Pie Chart Example")
plt.show()
```

```python
#creation of multipal plot at a time 
```

```python
x = np.linspace (0, 10,100)
y1 = np.sin(x)
y2 = np.cos(x)

#create subplot 
plt.figure(figsize=(10, 5))

#first subplot
plt.subplot(1, 2, 1)   #(Row, Column ,Index)
plt.plot(x, y1, 'r--')
plt.title("sin(x)")

#second subplot
plt.subplot(1, 2, 2)   #(Row, Column ,Index)
plt.plot(x, y2, 'b-')
plt.title("cos(x)")

plt.suptitle("Multiple Plot Example")  #overall title
plt.tight_layout()
plt.show()
```

```python
# Save a Plot

x = np.linspace(0, 5, 100)
y = np.exp(x)

plt.plot(x, y, 'm-')
plt.title("Exponential Function")
plt.xlabel("X")
plt.ylabel("exp(X)")

# Save plot as PNG file
plt.savefig("Matplotlib-Library-exponential_plot.png", dpi=300, bbox_inches='tight')

plt.show()
```

```python
# Create figure and subplots
fig, axs = plt.subplots(2, 2, figsize=(10, 8))

# Line plot
x = np.linspace(0, 10, 100)
axs[0, 0].plot(x, np.sin(x), 'r')  # 'r' = red
axs[0, 0].set_title("Sine Wave")

# Bar chart
categories = ['A', 'B', 'C', 'D']
values = [10, 15, 7, 12]
axs[0, 1].bar(categories, values, color='teal')

# Histogram
data = np.random.randn(1000)
axs[1, 0].hist(data, bins=25, color='purple', edgecolor='black')
axs[1, 0].set_title("Histogram")

# Scatter plot
x = np.random.rand(50)
y = np.random.rand(50)
axs[1, 1].scatter(x, y, color='orange')
axs[1, 1].set_title("Scatter Plot")

# Add main title for the whole figure
plt.suptitle("Matplotlib Dashboard Example", fontsize=14)
plt.tight_layout()

# Save plot as PNG file
plt.savefig("Matplotlib-Dashboard.png", dpi=300, bbox_inches='tight')

plt.show()
```
