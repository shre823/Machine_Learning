```python
# Array Indexing and Slicing

arr = np.array([10, 20, 30, 40, 50, 60])
print("\nOriginal Array:\n", arr)
print("\nFirst Element:\n", arr[0])
print("\nLast Element:\n", arr[-1])
print("\nSlice [1:4]:\n", arr[1:4])
print("\nEvery 2nd Element:\n", arr[::2])
print("\nEvery 2nd Element:\n", arr[1::2])
```

```python
# 2D Array slicing
arr2d = np.array([[10, 20, 30], [40, 50, 60], [70, 80, 90]])
print("\n2D Array:\n", arr2d)
print("\nElement at [1,2]:\n", arr2d[1, 2])
print("\nFirst two rows and first two columns:\n", arr2d[:2, :2])
```

```python
# Array Operations (Element-wise)

a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

print("\nAddition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
print("Power:", a ** 2)
```

```python
# Statistical and Mathematical Functions

arr = np.array([10, 20, 30, 40, 50, 100])

print("\nSum:", np.sum(arr))
print("Mean:", np.mean(arr))
print("Median:", np.median(arr))
print("Standard Deviation:", np.std(arr))
print("Minimum:", np.min(arr))
print("Maximum:", np.max(arr))
```

```python
# Reshaping and Flattening Arrays

arr = np.arange(1, 13)
print("\nOriginal 1D Array:", arr)

reshaped = arr.reshape(3, 4)
print("Reshaped to 3x4 Matrix:\n", reshaped)

# reshaped1 = arr.reshape(5, 5)
# print("Reshaped to 3x4 Matrix:\n", reshaped1)

flattened = reshaped.flatten()
print("Flattened Back to 1D:", flattened)
```

```python
# Concatenation and Stacking

x = np.array([[1, 2], [3, 4]])
y = np.array([[5, 6], [7, 8]])

print("\nHorizontal Stack:\n", np.hstack((x, y)))
print("Vertical Stack:\n", np.vstack((x, y)))
```

```python

# Boolean Indexing and Filtering

arr = np.array([5, 10, 15, 20, 25, 30])
filter_arr = arr > 15
print("\nFilter Boolean Array:", filter_arr)
print("Filtered Values (arr > 15):", arr[filter_arr])
```

```python
import numpy as np

# Linear Algebra Operations

mat1 = np.array([[1, 2], [3, 4]])
mat2 = np.array([[5, 6], [7, 8]])

print("\nMatrix 1:\n", mat1)
print("Matrix 2:\n", mat2)
print("\nMatrix Multiplication:\n", np.dot(mat1, mat2))
print("Transpose of Matrix 1:\n", mat1.T)
print("Determinant of Matrix 1:", np.linalg.det(mat1))
print("Inverse of Matrix 1:\n", np.linalg.inv(mat1))
```

```python
# Copy vs View (memory concept)

original = np.array([1, 2, 3, 4])
view_arr = original.view() # Creates a view: shares the same data buffer
copy_arr = original.copy() # Creates a copy: independent data buffer

original[0] = 99 # Modify the original array

print("\nOriginal Array:", original)
print("View Array (Changes reflected):", view_arr)
print("Copy Array (Independent):", copy_arr)


#view(): Creates a new array object that looks at the same data as original. No data is copied.
   # So when original[0] changes to 99, view_arr also shows 99 because both point to the same memory.
        
#copy(): Creates a completely new array with its own data. 
        #Changes to original do not affect copy_arr, which is why it still shows [1 2 3 4].
```
