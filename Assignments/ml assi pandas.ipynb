```python
# Pandas Assignment

# 1. Create a Pandas Series using a list of 5 random numbers and custom index names.
#     Then print the series and its data type.
import pandas as pd

# create series
Num = [10, 20, 30, 40, 50]
series = pd.Series(Num, index=['A', 'B', 'C', 'D', 'E'])
print(series)
print(series.dtype)

```

```python
# 2. Create a DataFrame with columns: "Student_Name", "Maths", "Science", "English".
#     Fill it with 5 sample student records and display the top 3 rows using head().
# create dataframe
df = pd.DataFrame({
    "Student_Name": ["Asha", "Ravi", "Priya", "Karan", "Neha"],
    "Maths": [85, 90, 78, 88, 92],
    "Science": [80, 95, 75, 85, 89],
    "English": [78, 88, 82, 90, 85]
})

# show top 3 rows
print(df.head(3))
```

```python
# 3. Import a CSV file (use any sample dataset) into a DataFrame.
#     Display the number of rows and columns using shape, and print column names.#only solve this 
# print rows and columns
print("Shape:", df.shape)

# print column names
print("Columns:", df.columns)
```

```python
# 4. Add a new column named "Total_Marks" that sums all three subject scores.
#     Then create another column "Average" by dividing Total_Marks by 3.
# add total marks
df["Total_Marks"] = df["Maths"] + df["Science"] + df["English"]

# add average
df["Average"] = df["Total_Marks"] / 3

print(df)
```

```python
# 5. Introduce a missing value (NaN) in the "Science" column and then:
#     (a) Identify missing values using isnull()
#     (b) Replace the missing value with the column mean using fillna()

# Introduce missing value in Science column
df.loc[2,  "Science"] = None
print("With Missing Data:\n", df)

# (a) Identify missing values
print(df.isnull())

# (b) Replace missing value with column mean
df["Science"] = df["Science"].fillna(df["Science"].mean())

print("\nAfter Filling Missing Value:\n", df)
```

```python
# 6. Filter and display all students whose "Average" score is greater than 75.
# Filter students with Average > 75
print(df[df['Average'] > 30])
```

```python
# 7. Sort the DataFrame by "Total_Marks" in descending order and show the top 3 performers.
# Sort by Total_Marks in descending order
top_3 = df.sort_values(by="Total_Marks", ascending=False).head(3)

print(top_3)
```

```python
# 8. Group students based on their "English" score range (like above or below 70)
#     and find the mean of "Maths" and "Science" marks for each group.
data = {
    'Student_Name': ['A', 'B', 'C', 'D', 'E'],
    'English': [65, 75, 80, 60, 72],
    'Maths': [70, 85, 90, 60, 88],
    'Science': [68, 78, 85, 55, 80]
}

df = pd.DataFrame(data)

# Create group based on English marks
df['English_Group'] = df['English'].apply(lambda x: 'Above 70' if x > 70 else '70 or Below')

# Group and calculate mean
grouped = df.groupby('English_Group')[['Maths', 'Science']].mean()

print(grouped)
```

```python
# 9. Merge two DataFrames:
#     (a) One with "Student_Name" and "Total_Marks"
#     (b) Another with "Student_Name" and "Sports_Score"
#     Then create a new column "Overall_Score" = Total_Marks + Sports_Score.
# First DataFrame
df1 = pd.DataFrame({
    "Student_Name": ["Asha", "Ravi", "Priya", "Karan", "Neha"],
    "Total_Marks": [243, 273, 235, 263, 266]
})

# Second DataFrame
df2 = pd.DataFrame({
    "Student_Name": ["Asha", "Ravi", "Priya", "Karan", "Neha"],
    "Sports_Score": [20, 25, 18, 22, 24]
})

# Merge both DataFrames
merged_df = pd.merge(df1, df2, on="Student_Name")

# Create Overall Score
merged_df["Overall_Score"] = merged_df["Total_Marks"] + merged_df["Sports_Score"]

print(merged_df)
```
